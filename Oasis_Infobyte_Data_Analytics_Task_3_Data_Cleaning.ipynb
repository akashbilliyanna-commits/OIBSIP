### Missing Data Handling

The `Peak` column contains 11 missing values and the `All Time Peak` column contains 14 missing values. These columns represent ranking information, so mean, median, or mode imputation would create artificial ranking values that are not supported by the dataset. Therefore, the missing values are retained as `NaN` rather than assigning inaccurate values. The remaining information in these rows is still useful for analysis.


```python
df['Peak'] = df['Peak'].astype('string').str.replace(r'\[.*?\]', '', regex=True)
```

### Standardisation and Data Type Correction

The dataset was standardised and the data types were corrected for analysis. Ranking values in the `Peak` and `All Time Peak` columns were cleaned by removing reference markers such as `[4]` and converted to numeric values. Monetary columns such as `Actual gross`, `Adjusted gross`, and `Average gross` were cleaned by removing currency symbols, commas, and reference markers, and converted to `float64`. The `Year(s)` column was kept as a string because it contains both single years and year ranges. Text columns were cleaned by removing unnecessary leading and trailing spaces. The `Ref.` column was stored as a string because it contains reference information.

After correction, the dataset contains appropriate data types for further analysis.


```python
Q1 = df['Shows'].quantile(0.25)
Q3 = df['Shows'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df[(df['Shows'] < lower) | (df['Shows'] > upper)]
```

```python
Q1 = df['Average gross'].quantile(0.25)
Q3 = df['Average gross'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df[(df['Average gross'] < lower) | (df['Average gross'] > upper)]
```

```python
Q1 = df['Actual\xa0gross'].quantile(0.25)
Q3 = df['Actual\xa0gross'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df[(df['Actual\xa0gross'] < lower) | (df['Actual\xa0gross'] > upper)]
```

```python
Q1 = df['Adjusted\xa0gross (in 2022 dollars)'].quantile(0.25)
Q3 = df['Adjusted\xa0gross (in 2022 dollars)'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df[(df['Adjusted\xa0gross (in 2022 dollars)'] < lower) | (df['Adjusted\xa0gross (in 2022 dollars)'] > upper)]
```

### Outlier Detection and Handling

The IQR (Interquartile Range) method was used to identify potential outliers in the numerical columns. Outliers were found in `Shows`, `Average gross`, `Actual gross`, and `Adjusted gross`. These observations represent unusually high values but are valid records in the dataset rather than obvious data-entry errors. Therefore, the identified outliers were retained and were not removed or capped.
